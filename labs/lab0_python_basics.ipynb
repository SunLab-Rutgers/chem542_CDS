## Importing a package
The Python ecosystem provides many useful packages, so you do not need to write every function you need to use. The imported package is usually called a **module**.

For example, if you would like to evaluate cos(x), you can simply import the `math` package and use the `cos` function in it, in the following manner:
```python
import math
y = math.cos(x)
```

You see that you need the dot "." after the module you imported.

You could also give aliases to a module by `import [module] as [module-alias]`, e.g.
```python
import math as ma
y = ma.cos(x)
```

If you only need specific functions in a module, you do not need to import the whole package, instead
```python
from math import cos
y = cos(x)
```

## Assigning values

Assigning values is straight forward. Suppose you want to give the variable `x` the value `5`, simply type
```python
x = 5
```

And then you can use `x` in the following lines knowing it's value is `5`, e.g.
```python
y = math.cos(x)
```
Then you are assigning the variable `y` with the value of `cos(5)`.

Using the `=` sign, you can assign anything to any variables.

**Caution**:
1. DO NOT give your variables common names. For example, you imported the module `math`, but then you define `math = 5`. Then the module `math` you imported earlier will be overwritten by `5`. Also, avoid names such as `list`, `str`, because these are predefined function names in Python.

2. Start your variable with a letter, upper case or lower case. You cannot start it with a number or special symbols. I would recommend give your variable an explicit name, so you do not forget what it is later.

- Good variable names: `molecule_name`, `property1`.

- Not so good variable names: `a`, `b` (what do they mean?)

- Variable names that won't work: `1a` (started with a number), `a-b` (the minus sign means minus!, use underscore), `list` (will work, but this is already taken).

## Printing
To print something, use the `print()` function in Python, e.g.
```python
print("Hello world!")

a = 5
print(a)
```

You could print many things together
```python
a = 5
b = 6
print("a = ", a, "b = ", b)
```

You can have formatted printing using the combination of `f` in front of a string and curly brackets `{}` enclosing the variable. The value of the variable will be printed.
```python
a = 5
print(f"a = {a}")
```
The above will print "a = 5".

If you want to be more precise, e.g., print 5.00, not 5.0000, you could use the following lines
```python
a = 5
print(f"a = {a:0.2f}")
```
where `:0.2f` means floating point number with 2 decimal places.

````python
# example
a = 5
print(f"a = {a:0.2f}")
````

## Defining and calling a function
An example of defining a python function:
```python
def summation_function(arg1, arg2):
    result = arg1 + arg2
    return result
```
Keys:
1. The `def` flag in front of the function name.
2. Use parenthesis to include arguments.
3. Remember the colon sign after the parenthesis
4. Indentation after the `def` line. Once the function is defined, you should remove the indentation in the next line. Otherwise, the next line still belongs to the definition of this function.
5. The value that a function returns is indicated by the `return` flag.

Some functions do not return anything, so you do not need the `return` line.

Calling a function is straightforward
```python
result = summation_function(arg1, arg2)
```

## List
A list is a very commonly used data type in Python, as well as in this course. It is indicated by a bracket `[]` enclosing items separated by comma `,`, The syntax is
```python
a_list = [item1, item2, item3]
```
Note that in computer language, we start counting from `0`, i.e., the 3rd item in a list has index `2` not `3`.

To get the 3rd item in the above list, use
```python
target_item = a_list[2]
```

To add a new member to a list, use the `append()` function:
```python
a_list = [item1, item2, item3]
a_list.append(item4)
```

An empty list is simply indicated as `[]`, combining the `append()` function, we could add an item to an empty or non-empty list.
```python
empty_list = []
empty_list.append(new_value)
```


## Dictionary
Another commonly used data type in data science. A dictionary is like a list, but with more information of each item value. It is composed of a curly bracket `{}` enclosing the key and value pairs (`key:value`), with pairs separated by comma `,`. E.g.

```python
a_dict = {key1: value1, key2: value2, key3: value3}
```
Now that we have specific indices for each pair, i.e. the key, we can find out the values by
```python
target_value = a_dict[key2]
```
which will assign `target_value` with `value2`.

If you want to see all keys or values, use `a_dict.keys()` or `a_dict.values()`.

You can also group them together by `a_dict.items()`.

An empty dictionary is simply `{}`, to assign a new value to a dictionary, use the following lines
```python
a_dict[new_key] = new_value
```

So you do not need an `append()` function.

Note that if the `new_key` already exists in the dictionary, then you are overwritting it's value with `new_value`.

````python
# example of lists and dictionaries
a_list = [1, 2, 3]
a_dict = {"k1": 1, "k2": 2, "k3": 3}

print("The second item in the list is ", a_list[1])
print("The value of the second item in the dictionary is ", a_dict["k2"])

print("Keys of the dictionary are: ", a_dict.keys())
print("Values of the dictionary are: ", a_dict.values())
print("Items of the dictionary are: ", a_dict.items())

a_list.append(4)
print("The new list is ", a_list)
````

## For loop
The last important concept is for loops. This allows us to apply the same operations to a set of targets.

We first need an iteratable variable, e.g., a list a dictionary. Then we can iterate over all members in it, e.g.
```python
a_list = [1, 2, 3]
for i in a_list:
    print(i)
```
Or
```python
a_dict = {"k1": 1, "k2": 2, "k3": 3}
for k in a_dict.keys():
    print(a_dict[k])
```

You are seeing the same pattern, the combination of colon `:` and indentation. This means that the indented part is under the condition of the line with the colon.

### Constructing a list with for loop
You could construct a list easily with a for loop. For example, we want to put the twice of the value of the list items in a new list, you can use a for loop as following
```python
a_list = [1, 2, 3]
a_list_double = [] # an empty list
for i in a_list:
    a_list_double.append(i * 2)
```

We could use a simpler way to replace the above
```python
a_list = [1, 2, 3]
a_list_double = [i*2 for i in a_list]
```
### Constructing a dictionary with for loop
You could also do the same thing to a dictionary,
```python
a_dict = {"k1": 1, "k2": 2, "k3": 3}
a_dict_double = {}
for k in a_dict: # when you loop over a dictionary, you are looping over the keys
    a_dict_double[k] = a_dict[k] * 2
```
replaced by
```python
a_dict = {"k1": 1, "k2": 2, "k3": 3}
a_dict_double = {k: v*2 for k, v in a_dict.items()}
```
