```python
from pymatgen.core import Structure, Lattice
```

## 1. Load CIF with pymatgen

```python
structure = Structure.from_file("caco3.cif", primitive=False) # primitive cell is the smallest cell, and in our case it is not
```

## Look at the composition
You can see that our unit cell contains 3 times of the component compared to a primitive cell:

CaCO3 -> Ca6C6O18

```python
comp = structure.composition
print(comp.formula)
```

### Look at lattice dimension
(a, b, c) and ($\alpha, \beta, \gamma$)

```python
lattice = structure.lattice

print(lattice.a, lattice.b, lattice.c)
print(lattice.alpha, lattice.beta, lattice.gamma)
```

## Look at the atomic positions

```python
coords = structure.cart_coords
for site in structure:
    element = site.species_string
    x, y, z = site.coords
    print(element, x, y, z)
```

## 2. Saving your own cif file with pymatgen

```python
# Define lattice (example)
lattice = Lattice.cubic(10.0)

# Cartesian coordinates
species = ["C", "C"]
coords = [
    [0, 0, 0],
    [1.2, 1.2, 1.2]
]

# Tell pymatgen these are Cartesian
structure = Structure(lattice, species, coords, coords_are_cartesian=True)

# Write CIF
structure.to(filename="output.cif")
```

3. Read a CIF file with ASE

```python
from ase.io import read
```

```python
atoms = read("caco3.cif")
```

### Look at the cell information
ASE uses the lattice vector,

$$\mathbf{L} = (\mathbf{a}, \mathbf{b}, \mathbf{c})$$


```python
cell = atoms.get_cell()
print(cell)
```

### Look at the atomic positions

```python
symbols = atoms.get_chemical_symbols()
coords = atoms.get_positions()
```

```python
for s, (x, y, z) in zip(symbols, coords):
    print(f"{s} {x:.6f} {y:.6f} {z:.6f}")
```
